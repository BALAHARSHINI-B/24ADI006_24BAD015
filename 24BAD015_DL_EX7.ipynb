```python
print("Roll No: 24BAD015 - BALA HARSHINI B")
import nltk
import string
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
sentences = brown.sents(categories='news')
original_text = [' '.join(sentence) for sentence in sentences]
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
processed_sentences = []
for sentence in original_text:
    words = word_tokenize(sentence.lower())
    words = [word for word in words if word not in string.punctuation]
    words = [word for word in words if word.isalpha() and word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]

    if words:
        processed_sentences.append(words)

print("\nOriginal Text:")
print(original_text[0])
print("\nPreprocessed Text:")
print(' '.join(processed_sentences[0]))
print("\nTotal Sentences:", len(original_text))
print("Preprocessed Sentences:", len(processed_sentences))
```

```python
print("Roll No: 24BAD015 - Bala Harshini B")
from nltk import FreqDist
from collections import Counter
all_words = [word for sentence in processed_sentences for word in sentence]
frequency = FreqDist(all_words)
vocabulary = set(all_words)
print("\nVocabulary Size:", len(vocabulary))

print("\nTop 10 Most Frequent Words:")
for word, count in frequency.most_common(10):
    print(word, ":", count)
print("\nTotal Word Occurrences:", len(all_words))
print("Unique Words:", len(vocabulary))
print("\nSample Preprocessed Sentences:")
for sentence in processed_sentences[:5]:
    print(sentence)
```

```python
print("Roll No: 24BAD015 - Bala Harshini B")
%pip install -q gensim
from gensim.models import Word2Vec
model = Word2Vec(
    sentences=processed_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=4,
    epochs=10,
    seed=42
)
print("\nWord2Vec Model Training Completed")
print("Vocabulary Size:", len(model.wv))
print("Embedding Dimensions:", model.wv.vector_size)
selected_word = next(
    word for word in ["government", "people", "state", "year", "american"]
    if word in model.wv
)

print("\nSelected Word:", selected_word)
print("\nEmbedding Vector:")
print(model.wv[selected_word])
print("\nVector Shape:", model.wv[selected_word].shape)

print("\nFive Most Similar Words:")
for word, score in model.wv.most_similar(selected_word, topn=5):
    print(word, ":", round(score, 4))
```

```python
print("Roll No: 24BAD015 - Bala Harshini B")
from itertools import combinations
candidate_pairs = [
    ("government", "president"),
    ("people", "country"),
    ("school", "student"),
    ("city", "state"),
    ("american", "national"),
    ("year", "month")
]

valid_pairs = [
    pair for pair in candidate_pairs
    if pair[0] in model.wv and pair[1] in model.wv
]
print("\nWord Pair Similarity Scores")

if len(valid_pairs) < 3:
    words = list(model.wv.index_to_key[:30])
    valid_pairs = [
        pair for pair in combinations(words, 2)
        if pair[0] in model.wv and pair[1] in model.wv
    ][:3]

results = []
for word1, word2 in valid_pairs:
    score = model.wv.similarity(word1, word2)
    results.append((word1, word2, score))
    print(word1, "-", word2, ":", round(float(score), 4))

if results:
    most_similar = max(results, key=lambda item: item[2])
    print("\nMost Similar Pair:")
    print(most_similar[0], "-", most_similar[1])
    print("Cosine Similarity:", round(float(most_similar[2]), 4))
    print("\nAnalysis:")
    print("Higher cosine similarity indicates that the words have")
    print("more similar contextual representations in the trained model.")
```
